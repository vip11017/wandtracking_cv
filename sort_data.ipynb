```python
import os
from pathlib import Path
import shutil
import csv
import pandas as pd
```

```python
data_dir = Path('data')
target_dir = Path('sorted_data')
target_dir.mkdir(exist_ok=True)
```

```python
for user in data_dir.iterdir():
    if user.is_dir():
        img_path = Path(f'{user}/images')
        if img_path.is_dir():
            for shape_folder in img_path.iterdir():
                if shape_folder.is_dir():
                    for img_file in shape_folder.iterdir():
                        if img_file.is_file():
                            file_name = f'{shape_folder.name}_{img_file.name}'
                            shutil.copy(img_file, target_dir / file_name)
```

```python
labels = {'ellipse', 'triangle', 'rectangle', 'other'}
file = Path("labels.csv")

with file.open("w", newline="") as f:
    writer = csv.writer(f)
    for file in target_dir.iterdir():
        if file.is_file() and file.suffix in {".png", ".jpg"}:
            shape_name = file.name.split("_")[0]
            if shape_name in labels:
                writer.writerow([file.name, shape_name])
            else:
                print(shape_name)
    f.close
```
